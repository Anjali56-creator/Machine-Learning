Python supports 3 categories of data types:

```python
#integer
print(4)
print(1e309) #last value
```

```python
#float
print(4.6)
print(1.7e308)
```

```python
#boolean
print(True)
print(False)
```

```python
#complex
print(4+5j)
```

```python
#string
print("kolkata")
print('kolkata')
print("""kolkata""")
```

```python
#list
print([1,2,3,4,5])
```

```python
#tuple
print((1,3,4))
```

```python
#sets
print({1,2,3,4,5})
```

```python
#dict
print({"Name":"Anjali","Age":19,"Gender":"Female"})
```
