Literals
Literals = basic data values for variables.

Types in Python:

Numeric literals
String literals
Boolean literals
Special literals

1.Numeric Literals

```python
 #Integer Literals
a = 0b1010  # Binary
b = 100     # Decimal
c = 0o310   # Octal
d = 0x12c   # Hex

# Float Literals
f1 = 10.5
f2 = 1.5e2
f3 = 1.5e-3

# Complex Literal
x = 3.14j

print(a, b, c, d)
print(f1, f2, f3)
print(x, x.imag, x.real)
```

2.String literals

```python

string = 'This is Python'
strings = "This is Python"
char = "C"
multiline_str = """This is a multiline string with more than one line code."""
unicode = u"\U0001f600\U0001F606\U0001F923"
raw_str = r"raw \n string"

print(string)
print(strings)
print(char)
print(multiline_str)
print(unicode)
print(raw_str)
```

Comparison Operators

```python
print(x > y)

print(x < y)

print(x >= y)

print(x <= y)

print(x == y)

print(x != y)
```

Logical Operators

```python
x = True
y = False
print( x or y)

print(x and y)

print(not x)
print(not y)
```

Bitwise Operators

```python
x = 2
y = 3
print(x & y) # AND

print(x | y) # OR

print(2 ^ 3) # XOR

print(x >> 2) # Right Shift

print(y << 3) # Left Shift

print(~ x) # NOT
```

Assignment Operators

```python
a = 3
print(a)

a += 3   # a = a + 3
print(a)

a -= 3   # a = a - 3
print(a)

a *= 3   # a = a * 3
print(a)

a /= 3   # a = a / 3
print(a)
```


Identity Operators

```python

a = 3
b = 3
print(a is b)

a = "Hello"
b = "Hello"
print(a is b)

a = [1, 2, 3]
b = [1, 2, 3]
print(a is b)

a = "Hello-World"
b = "Hello-World"
print(a is b)
print(a is not b)
```


Membership Operators
in & not in
