```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.xlabel('year')
plt.ylabel('Runs scored')
plt.title("Rohit sharma vs Virat kohli")
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='pink')
plt.plot(batsman['index'],batsman['RG Sharma'],color='purple')
plt.xlabel('year')
plt.ylabel('Runs scored')
plt.title("Rohit sharma vs Virat kohli")
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='pink',linestyle='dashed',linewidth='2')
plt.plot(batsman['index'],batsman['RG Sharma'],color='purple',linestyle='solid',linewidth='4')
plt.xlabel('year')
plt.ylabel('Runs scored')
plt.title("Rohit sharma vs Virat kohli") 
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='pink',linestyle='dashed',marker='D',linewidth='2')
plt.plot(batsman['index'],batsman['RG Sharma'],color='purple',linestyle='solid',marker='o',linewidth='3')
plt.xlabel('year')
plt.ylabel('Runs scored')
plt.title("Rohit sharma vs Virat kohli") 
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='pink',linestyle='dashed',marker='D',linewidth='2',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='purple',linestyle='solid',marker='o',linewidth='3',label='Rohit')
plt.xlabel('year')
plt.ylabel('Runs scored')
plt.title("Rohit sharma vs Virat kohli") 
plt.legend() #legend default value is best i.e; it occupy the best position by itself.
```
